# 🌟 STEP 5: 상권 유형 분류 (Clustering)

이 노트북은 `merged_features_v3.csv`를 기반으로
서울시 상권을 **주거형 / 업무형 / 관광형 / 내수형**으로 분류합니다.

### 사용 변수
- visitor_ratio
- short_foreigner_ratio
- visit_intensity
- day_ratio


```
# 1. 라이브러리 로드
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
```

```
# 2. 데이터 로드
df = pd.read_csv('merged_features_v3.csv')
features = ['visitor_ratio','short_foreigner_ratio','visit_intensity','day_ratio']
X = df[features].dropna()
X.head()
```

```
# 3. 스케일링
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```
# 4. Elbow Method
inertia = []
for k in range(2,8):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.plot(range(2,8), inertia, marker='o')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

➡️ Elbow 결과를 기준으로 **k=4**를 선택합니다.

```
# 5. KMeans Clustering (k=4)
kmeans = KMeans(n_clusters=4, random_state=42)
df_clustered = df.loc[X.index].copy()
df_clustered['cluster'] = kmeans.fit_predict(X_scaled)
df_clustered[['cluster'] + features].head()
```

```
# 6. 군집별 평균 특성
cluster_summary = df_clustered.groupby('cluster')[features].mean()
cluster_summary
```

```
# 7. 시각화
sns.pairplot(df_clustered, vars=features, hue='cluster', diag_kind='kde')
plt.show()
```
